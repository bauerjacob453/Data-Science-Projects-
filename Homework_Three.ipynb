Employee records from a large organization were gathered. There are two files:

- 'employee_salary.csv' contains salary and bonus information, as well as some additional demographics
- 'employee_team.csv' contains information about which team the employee is on, as well as if the employee is part of senior management.

1. Read the salary dataset into your notebook. Using this data, answer the following questions.
   - How many observations are in the dataset? What types of variables are available upon read-in?
   - How many missing values are present in each variable?
   - Convert the StartDate variable to a datetime object
   - Remove duplicate entries. How many duplicate entries were removed? How can you tell?
   - How many of these employees are named 'Robert'?

```python
import pandas as pd
from matplotlib import pyplot as plt 
import seaborn as sns 
import numpy as np
employee_salary_df = pd.read_csv("employee_salary.csv")
employee_team_df = pd.read_csv('employee_team.csv')
print(f"The length of the employee salary dataframe is: {len(employee_salary_df)}")
print(employee_salary_df.info()) # The data types are: int, object, and float
missing_salary = employee_salary_df.isna()
missing_salary_count = missing_salary.sum()
print(missing_salary_count)
employee_salary_df["StartDate"] = pd.to_datetime(employee_salary_df["StartDate"])
employee_salary_df = employee_salary_df.drop_duplicates()
print(f"The new length of the employee salary dataframe is: {len(employee_salary_df)}") # 35 duplicates were removed. Original df = 1035. New df = 1000. 1035-1000 = 35
employee_salary_df["FirstName"] == employee_salary_df["FirstName"].str.strip().str.lower().str.capitalize()
robert_count = (employee_salary_df["FirstName"] == "Robert").sum()
print(f"There are {robert_count} employees named Robert.")
```

5. Use KNN imputation to fill in the missing Salary values. Explain your choices.

```python
from sklearn.impute import KNNImputer
imputer = KNNImputer(n_neighbors = 100) # I made n_neighbors = 100 to ensure that the imputer was gathering enough data to make a valid assumption of what the salary of that person would be
merge_df_imputed = merge_df.copy() # I made a copy of the data to not change the initial dataframe from the imputations
merge_df_imputed["Salary"] = imputer.fit_transform(merge_df[["Salary", "BonusPercent"]]) # I only chose the salary column because this is the column we are looking at
print("KNN imputed data: ", "\n", merge_df_imputed.head())
missing_merge = merge_df_imputed.isna()
missing_merge_count = missing_merge.sum()
print(missing_merge_count) # Printed this to see if the KNN imputation worked properly
```
